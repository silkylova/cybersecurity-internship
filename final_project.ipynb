```python
"""
Day 15 Final Project: SE Attack Chain Simulator
Combines OSINT tracking, profile scoring, phish detection, template generation, and IR triggers.
"""
import datetime
import json
import re
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

MODULES = {
    "osint": "Run passive OSINT domain and email footprint check",
    "profile": "Build target behavioral risk score (Fake Profile Detector)",
    "phish": "Score an email/URL for phishing indicators (ML Naive Bayes)",
    "template": "Generate spear-phishing simulation training template",
    "ir": "Trigger automated Incident Response (IR) containment workflow"
}

def banner():
    print("\n" + "="*50)
    print("■■■ SE CHAIN SIMULATOR — FINAL PROJECT ■■■")
    print("■ Zhibek Mussakulova | Cybersecurity Internship ■")
    print("="*50)

def run_osint():
    print("\n--- [1] OSINT & DOMAIN FOOTPRINT MODULE ---")
    domain = input("Enter target domain (e.g., target-corp.com): ").strip()
    print(f"[*] Querying passive DNS and registration records for {domain}...")
    print(f"[+] Domain Active: {domain}")
    print(f"[+] Discovered Mail Exchange (MX): mail.{domain}")
    print(f"[+] Estimated Exposed Employee Footprint: 42 records found.")

def run_profile():
    print("\n--- [2] TARGET PROFILE RISK SCORER (Day 9 Heuristic) ---")
    age = int(input("Enter account age in days: "))
    following = int(input("Enter following count: "))
    followers = int(input("Enter follower count: "))
    has_pic = input("Has profile picture? (y/n): ").lower() == 'y'
    posts = int(input("Enter total post count: "))

    score = 0
    reasons = []
    if age < 30:
        score += 30
        reasons.append("Account under 30 days old (+30)")
    if followers > 0 and (following / followers) > 10:
        score += 25
        reasons.append("High following-to-follower ratio (+25)")
    if not has_pic:
        score += 20
        reasons.append("Missing profile picture (+20)")
    if posts < 5:
        score += 15
        reasons.append("Fewer than 5 posts (+15)")

    score = min(score, 100)
    verdict = "Likely genuine (0-29%)" if score < 30 else ("Suspicious (30-59%)" if score < 60 else "Likely fake/bot (60-100%)")

    print(f"\n[+] Risk Score: {score}% — Verdict: {verdict.upper()}")
    print("[-] Triggered Indicators:", reasons if reasons else "None")

def run_phish():
    print("\n--- [3] PHISHING ML CLASSIFIER (Day 12 Naive Bayes) ---")
    # Quick embedded training set for the demo tool
    train_emails = [
        ("Verify your account now or it will be suspended", 1),
        ("Click here to claim your prize immediately", 1),
        ("Team standup at 3pm, agenda attached", 0),
        ("Your invoice for Q2 is ready for review", 0),
        ("Urgent: update your bank details to avoid closure", 1),
        ("Meeting notes from yesterday", 0)
    ]
    X, y = zip(*train_emails)
    pipe = Pipeline([("vec", CountVectorizer()), ("clf", MultinomialNB())])
    pipe.fit(X, y)

    sample = input("Enter email text to analyze: ")
    pred = pipe.predict([sample])[0]
    print(f"\n[+] Classification Result: {'🚨 PHISHING DETECTED (1)' if pred else '✅ LEGITIMATE (0)'}")

def run_template():
    print("\n--- [4] SPEAR-PHISHING TRAINING TEMPLATE GENERATOR ---")
    target_name = input("Enter target employee name: ")
    department = input("Enter target department (e.g., HR, Finance): ")

    template = f"""
    SUBJECT: Urgent Review Required: Updated {department} Direct Deposit Form

    Dear {target_name},

    We are updating our payroll portal compliance records. Please verify your banking credentials immediately via the secure portal link below to avoid interruption in your upcoming pay cycle:

    https://internal-portal-secure-verify.com/login?user={target_name.lower().replace(' ', '')}

    Sincerely,
    Human Resources Operations
    """
    print("\n[Generated Training Template]:")
    print(template)

def run_ir():
    print("\n--- [5] INCIDENT RESPONSE AUTOMATION (Day 14 Workflow) ---")
    user = input("Enter compromised user email: ")
    severity = input("Enter incident severity (MEDIUM/HIGH/CRITICAL): ").upper()

    actions = []
    if severity in ('HIGH', 'CRITICAL'):
        actions = [
            f"LOCK account: {user}",
            f"REVOKE active session tokens for {user}",
            "NOTIFY SOC emergency channel",
            "PRESERVE mail and auth logs for forensics"
        ]
    else:
        actions = [f"FLAG user {user} for monitoring"]

    print(f"\n[+] Triggering IR Playbook for {user} [{severity}]...")
    for a in actions:
        print(f"   [x] {a}")

    report = {"user": user, "severity": severity, "actions": actions, "timestamp": datetime.datetime.now().isoformat()}
    with open(f"final_ir_report_{user.split('@')[0]}.json", "w") as f:
        json.dump(report, f, indent=4)
    print(f"[-] Incident report saved to disk.")

def menu():
    while True:
        banner()
        for k, v in MODULES.items():
            print(f" [{k}] {v}")
        print(" [exit] Quit simulator")

        choice = input("\nSelect module: ").strip().lower()
        if choice == 'exit':
            print("Exiting SE Attack Chain Simulator. Stay safe!")
            break
        elif choice in MODULES:
            print(f"\n[+] Launching {choice} module...")
            if choice == 'osint': run_osint()
            elif choice == 'profile': run_profile()
            elif choice == 'phish': run_phish()
            elif choice == 'template': run_template()
            elif choice == 'ir': run_ir()
            input("\nPress Enter to return to main menu...")
        else:
            print("Invalid choice. Please select a valid module key.")

if __name__ == "__main__":
    menu()
```
