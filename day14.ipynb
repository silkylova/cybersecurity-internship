```python
"""
Day 14: SE Incident Response Plan
Automated IR workflow script for containment actions.
"""
import datetime
import json

def ir_response(incident):
    print(f"\n{'='*50}")
    print(f"🚨 INCIDENT RESPONSE TRIGGERED")
    print(f"{'='*50}")
    print(f"Time     : {datetime.datetime.now().isoformat()}")
    print(f"Type     : {incident['type'].upper()}")
    print(f"Severity : {incident['severity'].upper()}")
    print(f"User     : {incident['user']}")

    actions = []

    # 1. Base containment based on severity
    if incident['severity'] in ('HIGH', 'CRITICAL'):
        actions.extend([
            f"LOCK account: {incident['user']}",
            f"REVOKE all active sessions for {incident['user']}",
            "NOTIFY SOC team via emergency channel",
            "PRESERVE authentication and mail logs for forensics"
        ])
    elif incident['severity'] == 'MEDIUM':
        actions.extend([
            f"FLAG account {incident['user']} for enhanced monitoring",
            "NOTIFY user to verify recent activity"
        ])

    # 2. Specific eradication actions based on attack vector
    if incident['type'] == 'phishing':
        actions.extend([
            "QUARANTINE malicious email from all inboxes",
            "BLOCK sender domain at email gateway",
            "SCAN attachments in sandbox environment"
        ])
    elif incident['type'] == 'brute_force':
        actions.extend([
            "BLOCK source IP addresses at firewall",
            "FORCE password reset on next login"
        ])
    elif incident['type'] == 'suspicious_email_rule':
        actions.extend([
            f"DELETE unauthorized inbox rules for {incident['user']}",
            "SCAN mailbox for exfiltrated sensitive data"
        ])

    print("\nContainment & Eradication Actions Executed:")
    for a in actions:
        print(f" [x] {a}")

    # 3. Generate Compliance Report
    report = {
        "incident_metadata": incident,
        "actions_taken": actions,
        "timestamp": datetime.datetime.now().isoformat()
    }

    filename = f"ir_report_{incident['user']}.json"
    with open(filename, "w") as f:
        json.dump(report, f, indent=4)
    print(f"\n✅ IR report securely saved to: {filename}")


if __name__ == "__main__":
    # Simulate an automated SIEM alert triggering the response script
    ir_response({"type": "phishing", "severity": "HIGH", "user": "riya@sqrock.com"})

    # Simulate a critical post-compromise alert
    ir_response({"type": "suspicious_email_rule", "severity": "CRITICAL", "user": "admin@sqrock.com"})
```
